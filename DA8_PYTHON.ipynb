```python
#CONDITIONAL STATEMENTS
n = int(input())
if n <= 100:

  bill = n * 2
  print(bill)
elif n <= 200:
  bill = ((100*2) + (n - 100)*3)
  print(bill)
elif n <= 400:
  bill = (( 100 * 2) + (100 * 3) + (n - 200) * 3)
  print(bill)
else:
  bill = ((100 * 2) + (100 * 3) + (200 * 5) + (n - 400) * 7)
  print(bill)
if bill > 2000:
  bill = bill * 1.10
  print("with extra charges the final bill :",bill)
```
