leap year

```python
year = int(input())
if (year %400 == 0):

  print("leap year")
elif (year%4==0 and year%100 !=0):
  print("leap year")
else:
  print("not a leap year")
```

mail login

```python
user_name = str(input())
password = str(input())
u = input()
p = input()
if user_name == u and password == p:
  print("login successful")
elif user_name != u and password == p:
  print("invalid user_name")
elif user_name == u and password != p:
  print("invalid password")
else:
  print("login failed")
```
