```python
#calculator n number of operators
a = int(input())
b = int(input())
n = int(input())
for i in range(n):
   operator = input()
   if operator == '+':
    print(a + b)
   elif operator == '-':
    print(a -b)
   elif operator == '*':
    print(a * b)
   elif operator == '/':
    print( a / b)
   elif operator == '**':
    print(a ** b)
   elif operator == '//':
    print(a // b)
   else:
    print("not a valid operator")
```
